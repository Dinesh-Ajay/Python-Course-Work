```python
import numpy as np

arr1 = np.array([1,2,3,4,5])
print('1-dim',arr1,sep='\n',end='\n\n')

arr2 = np.array([[1,2,3],[4,5,6]])
print('2-dim',arr2,sep='\n',end='\n\n')

arr3 = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print('multi-dim',arr3,sep='\n')
```

```python
z = np.zeros((3,4))
print(z)
```
