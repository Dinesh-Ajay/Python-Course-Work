```python
import matplotlib.pyplot as plt

hours=[1,2,3,4,5,6,7,8]
visitors=[50,60,65,70,90,120,110,130]
plt.plot(hours,visitors,
         color="red",
         marker="o",
         linestyle="-",
         linewidth=1,
         alpha=1,
         label="Visitors")
plt.title("website visitors Per Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Visitors")
plt.legend()
plt.grid(True,linestyle="-",alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt
products=["Laptop","Mobile","Table","Headphones","Smartwatch"]
sales=[250,400,150,300,180]
plt.bar(products,sales,color="green",edgecolor="red",alpha=0.5,width=0.5)
plt.title("Product Sales Comparision")
plt.xlabel("Products")
plt.ylabel("Units sold")
plt.show()
```

```python
import matplotlib.pyplot as plt

brands=["Apple","Samsumg","Xiaomi","Oppo","Others"]
market_share = [30,25,20,10,15]
plt.pie(market_share,
        labels = brands,
        autopct="%.2f%%",
        startangle=0,
        colors=['gold','skyblue','lightcoral','lightgreen','violet'],
        explode = [0.1,0,0,0,0],
        shadow=True)
plt.title("Mobile Market Share")
plt.show()
```

```python
import matplotlib.pyplot as plt

scores = [
    [70,86,90,92,88,76,95,9],
    [65,70,68,72,75,80,85,78],
    [90,92,94,96,88,91,89,93],
    [62,60,61,90,65,63,67,61],
    [70,8,58,50,92,91,100,89]
]
plt.boxplot(scores,
            patch_artist=True,
            medianprops=dict(color='red',linewidth=2),
            boxprops=dict(facecolor='lightblue'),
            whiskerprops=dict(color='green'),
            capprops=dict(color='black'),
            notch=False,
            vert=True,
            labels=['A','B','C','D','E']
            )
plt.title("Exam Scores Distribution Across Calsses")
plt.xlabel("Classes")
plt.ylabel("Scores")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
