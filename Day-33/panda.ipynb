```python
import pandas as pd

prices = [2999,15999,52999,4999,1999]
products=['Wireless Earbuds', 'Smartphone', 'Laptop','Smartwatch', 'Bluetooth Speaker']

product_price = pd.Series(prices, index=products)
print(product_price)
```

```python
print("Mean: ", product_price.mean())
print("Sum: ", product_price.sum())
print("Max: ", product_price.max())
print("Min: ", product_price.min())
print("Median: ", product_price.median)
```

```python
print("Head (First 3 elements):\n", product_price.head(3))
print("Tail (Last 3 elements):\n", product_price.tail(3))
```

```python
print("Apply (Adding 18% GST):\n",product_price.apply(lambda x: f'₹{x+(x * 0.18)}'))
```

```python
print("Map (Formating as Currency):\n",product_price.map(lambda x: f'₹{x:.2f}'))
```

```python
print(product_price.sort_values())
print(product_price.sort_values(ascending=False))
```

```python
print(product_price.sort_index())
print(product_price.sort_index(ascending=False))
```

```python
print("Value Counts:\n",product_price.value_counts())
```

```python
import pandas as pd

data = {
    "Product":['Wireless Earbuds', 'Smartphone', 'Laptop','Smartwatch', 'Bluetooth Speaker'],
    'Brand':['Sony', 'Apple', 'Dell', 'Fitbit', 'JBL'],
    "Price":[2999,15999,52999,4999,1999],
    'Stock':[50,30,20,40,60],
    'Bestseller':[True, False,False,True,True]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.shape)
print(df.columns)
print(df.info())
```

```python
print(df.head(3))
print('\n',df.tail(3))
```

```python
print(df)
print('\n1st Product:',df.loc[1,"Product"])
print('\n4th Product:',df.loc[4,"Product"])
print('\n4th Stock:',df.loc[4,"Stock"])

print('\n',df.iloc[4,0])
print('\n',df.iloc[2,1])
print('\n',df.iloc[3,3])
```

```python
print(df.iloc[3])
print(df.iloc[2])
```

```python
print(df)
print("\n*********************Using loc*********************",df.loc[df['Bestseller']==True])
print("\n*********************Using loc*********************",df.loc[df['Stock']<40])
print("\n*********************Using loc*********************",df.loc[df['Price']<10000])
```

```python
df_renamed = df.rename(columns={"Price":"Cost"})
print(df_renamed)
```

```python
df_grouped = df.groupby("Brand").agg({"Price":"sum", "Stock":"sum"})
print(df_grouped)
```

```python
grouped = df.groupby("Brand").agg({"Price":["mean", "max", "min"]})
print(grouped)
```

```python
grouped = df.groupby("Brand")["Price"].mean()
print(grouped)
```

```python
data2 = {
    'Brand':['Sony', 'Apple', 'Dell', 'Fitbit', 'JBL'],
    "Rating":[4.2,4.5,4.0,4.1,3.9],
    'Discount':[28,40,16,10,5]
}

df_rating = pd.DataFrame(data2)
print(df_rating)
```

```python
df_merged = df.merge(df_rating, on="Brand")
print("Merged DataFrame (Adding Ratings):\n",df_merged)
```

```python
df_merged = pd.merge(df, df_rating, on="Brand")
print("Merged DataFrame (Adding Ratings):\n",df_merged)
```

```python
new_date = {
    "Product":['Tablet'],
    "Brand":['Samsung'],
    "Price":[12999],
    "Stock":[25],
    "Bestseller":[True]
}
df_new = pd.DataFrame(new_date)
print(df_new)
```

```python
df_concat = pd.concat([df,df_new],ignore_index=True)
print("Concatenated DataFrame:\n",df_concat)
```
