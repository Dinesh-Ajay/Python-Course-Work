```python
import pandas as pd
import requests

from bs4 import BeautifulSoup
```

```python
# URL = "https://codegnan.com/"
URL = "https://timely-sunshine-e821b3.netlify.app/"
```

```python
page = requests.get(URL)
```

```python
page.status_code
```

```python
htmlCode = page.text
print(htmlCode)
```

```python
soup = BeautifulSoup(htmlCode, 'html.parser')
print(soup)
```

```python
content = soup.find_all('div', attrs={"class":"a"})
for i in content:
  print("Div:",i,end="\n\n\n\n")
```

```python
content = soup.find('div', attrs={'class':'name'})
text = content.text
print(text)
```
